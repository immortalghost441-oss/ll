```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("youtube_analytics.csv")
```

display(df.head())

df.info()

```python
df["upload_date"] = pd.to_datetime(df["upload_date"])
```

```python
top_videos = df.sort_values(
    by="views",
    ascending=False
)

top_videos[["video_title", "views"]]
```

```python
plt.bar(
    top_videos["video_title"],
    top_videos["views"]
)

plt.xticks(rotation=45, ha="right")

plt.title("Views by Video")
plt.xlabel("Video")
plt.ylabel("Views")

plt.tight_layout()
plt.show()
```

```python
import math
corr = df["video_length_minutes"].corr(
    df["watch_time_hours"]
)

print(f"Correlation: {corr:.2f}")
print('correlation betwwen vid length and watch hours is',math.ceil(corr))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["video_length_minutes"],
    df["watch_time_hours"]
)

plt.xlabel("Video Length (minutes)")
plt.ylabel("Watch Time (hours)")
plt.title("Video Length vs Watch Time")

plt.show()
```
